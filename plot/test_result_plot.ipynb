```python
import json
import numpy as np 
from matplotlib import pyplot as plt
```

```python
# 描绘训练结果
file_head = '../Baselines/train/'
## e/all/loss.json
WarTransGRU_result = 'MSE.json'    
# 打开训练结果.json文件
with open(file_head + WarTransGRU_result,'r') as f:
    WarTransGRU_result_file = json.load(f)
```

```python
def smooth(data_np, n_point=20):
    last = np.ones(n_point-1)*data_np[0]
    data_np = np.concatenate((last,data_np),axis=0)
    smoothed = []
    for index in range(len(data_np)-n_point+1):
        
        smoothed_val = data_np[index:index+n_point].mean()
        smoothed.append(smoothed_val)
    return np.array(smoothed)
```

```python
result_MSE_mean, \
result_MSE, \
result_MAE_mean, \
result_RMSE_mean, \
result_STEP = smooth(np.array(WarTransGRU_result_file['record_MSE'])), \
                np.array(WarTransGRU_result_file['record_MSE']),\
                smooth(np.array(WarTransGRU_result_file['record_MAE'])),\
                smooth(np.array(WarTransGRU_result_file['record_RMSE'])), \
                np.array(WarTransGRU_result_file['record_step'])

print(len(result_STEP))
result_index = np.arange(len(result_STEP))
```

```python
fig = plt.figure()
# plt.title("Mean loss of training") 
plt.xlabel("step") 
plt.ylabel("MSE") 


plt.plot(result_STEP, result_MSE_mean[result_index],label='train_MSE', color = 'b',linewidth=1) 
# plt.plot(result_index, result_MAE_mean[result_index],label='train_MAE', linewidth=1) 
# plt.plot(result_index, result_RMSE_mean[result_index],label='train_RMSE', linewidth=1) 



plt.legend()
plt.axis([-50000, 1500000, 0, 0.045])
plt.show()
fig.savefig('MSE.png',format='png')
# fig.savefig('loss.eps',format='eps')
```

```python
WarTransGRU_result_path = '../Baselines/test/WarTransGRU/base'


for filename in os.listdir(WarTransGRU_result_path):
    file_path = os.path.join(WarTransGRU_result_path, filename)
    with open(file_path,'rb') as f:
        WarTransGRU_result_file = pickle.load(f)
    
    model_MSE_mean = []
    model_MAE_mean = []
    model_RMSE_mean = []

    for key, value in WarTransGRU_result_file.items():
        MSE_mean = np.mean(value['record_MSE'])
        MAE_mean = np.mean(value['record_MAE'])
        RMSE_mean = np.mean(value['record_RMSE'])
        model_MSE_mean.append(MSE_mean)
        model_MAE_mean.append(MAE_mean)
        model_RMSE_mean.append(RMSE_mean)

    print(filename)
    print("all game MSE/MAE/RMSE mean:", np.mean(model_MSE_mean), np.mean(model_MAE_mean), np.mean(model_RMSE_mean))
        
# with open(file_head + ELP_result,'rb') as f:
#     ELP_result_file = pickle.load(f)
```
